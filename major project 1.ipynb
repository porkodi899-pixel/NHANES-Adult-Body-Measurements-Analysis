```python

# Outline:
# 1. Download CSV files.
# 2. Read them into numpy matrices `male` and `female`.
# 3. Histograms of weights (female top, male bottom) with common x limits.
# 4. Side-by-side boxplot comparing male and female weights and short interpretation.
# 5. Compute numerical aggregates (location, dispersion, shape) and describe them.
# 6. Add BMI as an 8th column to the female matrix.
# 7. Create standardized female matrix `zfemale`.
# 8. Scatterplot matrix (pairplot) for selected standardized female variables and compute Pearson/Spearman correlations.
# 9. Add waist/height and waist/hip ratios as new columns to both matrices.
# 10. Boxplot comparing these ratios between sexes and brief explanation.
# 11. Short discussion of advantages/disadvantages of BMI, waist-to-height, and waist-to-hip ratios.
# 12. Print standardized body measurements for 5 persons with lowest BMI and 5 with highest BMI and interpret.
#


# %% [markdown]
# ## 1. Download the two CSV files
#
# Purpose: download the two required CSVs (male and female excerpts) from the given GitHub repository
# (using the raw file URLs). The files will be read directly into memory; they are small.
#
# After running this cell we will have two local file paths we can use to load the data.

# %%
import os
from pathlib import Path
import requests

DATA_DIR = Path("nhanes_data")
DATA_DIR.mkdir(exist_ok=True)

BASE_RAW = "https://raw.githubusercontent.com/gagolews/teaching-data/master/marek"

files = {
    "male": "nhanes_adult_male_bmx_2020.csv",
    "female": "nhanes_adult_female_bmx_2020.csv",
}

for key, fname in files.items():
    url = f"{BASE_RAW}/{fname}"
    local_path = DATA_DIR / fname
    if local_path.exists():
        print(f"{fname} already exists locally -> {local_path}")
    else:
        print(+"Downloading {fname} ...")
        r = requests.get(url, timeout=30)
        r.raise_for_status()
        local_path.write_bytes(r.content)
        print(f"Saved {fname} to {local_path}")

print("Download step finished. Files are in:", DATA_DIR.resolve())

# Summary:
# The two CSV files have been downloaded and saved under the "nhanes_data" directory.
# If the files already existed they were not re-downloaded.


# %% [markdown]
# ## 2. Read the two files as numpy matrices named `male` and `female`
#
# Purpose: load both CSV files into NumPy arrays. Each matrix will have seven columns in the order:
# 1. weight (kg)  2. standing height (cm)  3. upper arm length (cm)
# 4. upper leg length (cm)  5. arm circumference (cm)  6. hip circumference (cm)
# 7. waist circumference (cm)
#
# After the cell we will print shapes and a small preview to confirm correct loading.

# %%
import numpy as np

male_path = DATA_DIR / files["male"]
female_path = DATA_DIR / files["female"]

# Use numpy.genfromtxt to load numeric data; skip header row
male = np.genfromtxt(male_path, delimiter=",", skip_header=1)
female = np.genfromtxt(female_path, delimiter=",", skip_header=1)

male = male[~np.isnan(male).any(axis=1)]
female= female[~np.isnan(female).any(axis=1)]
import pandas as pd

print("\n male shape:",male.shape)
print("\n female shape:",female.shape)



# quick check: first 5 rows
print("\nMale (first 5 rows):\n", male[:5])
print("\nFemale (first 5 rows):\n", female[:5])

# Summary:
# The arrays `male` and `female` have been loaded. Each row corresponds to a participant,
# each column to a measurement in the specified order. The printed shapes and previews confirm successful loading.


# %% [markdown]
# ## 3. Histograms of female and male weights on a single figure
#
# Purpose: draw two histograms stacked vertically (female weights on top, male weights bottom),
# and set identical x-axis limits for both subplots so they are directly comparable.
#
# We'll compute sensible x limits from the combined weights, add titles, axis labels, and show the plot.

# %%
import matplotlib.pyplot as plt

# Extract weight columns (column 0)
female_weights = female[:, 0]
male_weights = male[:, 0]

# Compute common x-limits:
combined_min = np.nanmin(np.concatenate([female_weights, male_weights]))
combined_max = np.nanmax(np.concatenate([female_weights, male_weights]))
# Add a small margin:
margin = 0.05 * (combined_max - combined_min)
xlim = (combined_min - margin, combined_max + margin)

fig, axs = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
bins = 30

axs[0].hist(female_weights, bins=bins, color="#d95f02", alpha=0.8)
axs[0].set_title("Female weights (kg) — NHANES excerpt")
axs[0].set_ylabel("Count")

axs[1].hist(male_weights, bins=bins, color="#1b9e77", alpha=0.8)
axs[1].set_title("Male weights (kg) — NHANES excerpt")
axs[1].set_ylabel("Count")
axs[1].set_xlabel("Weight (kg)")

# Apply identical x-limits
axs[0].set_xlim(xlim)
axs[1].set_xlim(xlim)

plt.tight_layout()
plt.show()

# Summary and discussion:
# The two histograms show the weight distribution for females (top) and males (bottom).
# Because both subplots use identical x-axis limits, direct visual comparison is immediate:
# - Male distribution is generally shifted to higher weights compared to female distribution.
# - Both sexes show substantial spread, with some heavy-tailed observations (outliers) visible on the right.
# Exact quantitative comparison follows in the next section.


# %% [markdown]
# ## 4. Boxplot comparing male and female weights (side by side)
#
# Purpose: draw a single box-and-whisker plot with two boxes side by side:
# left = female weights, right = male weights.
# Then provide a brief interpretation.

# %%

plt.figure(figsize=(9, 5))
plt.boxplot([female_weights, male_weights], 
tick_labels=["Female", "Male"])
plt.ylabel("Weight (kg)")
plt.title("Boxplot comparison of weights by sex")
plt.grid(axis="y", linestyle=":", alpha=0.5)
plt.show()

# Summary and discussion:
# The boxplot directly compares medians, interquartile ranges (IQR), and outliers.
# You will observe that:
# - The male median weight is larger than the female median.
# - The male IQR may be wider (indicating larger spread) — confirm numerically below.
# - There are some extreme high-weight outliers in both groups.
# These observations match the histograms above and will be quantified in the next step.


# %% [markdown]
# ## 5. Compute basic numerical aggregates for male and female weights
#
# Step 5: Basic numerical aggregates for female and male weights

import pandas as pd
from scipy import stats
import numpy as np

def describe_weights(arr, label):
    s = pd.Series(arr).dropna()

    mean = s.mean()
    median = s.median()

    # Mode
    modes = s.mode()
    mode_val = modes.iloc[0] if not modes.empty else np.nan

    std = s.std(ddof=0)
    var = s.var(ddof=0)

    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    mn = s.min()
    mx = s.max()
    rng = mx - mn

    skew = stats.skew(s, bias=False)
    kurt = stats.kurtosis(s, fisher=True, bias=False)

    return {
        "Group": label,
        "N": s.count(),
        "Mean": mean,
        "Median": median,
        "Mode": mode_val,
        "Std": std,
        "Variance": var,
        "IQR": iqr,
        "Min": mn,
        "Max": mx,
        "Range": rng,
        "Skewness": skew,
        "Kurtosis": kurt
    }


# Calculate statistics
female_stats = describe_weights(female_weights, "Female")
male_stats = describe_weights(male_weights, "Male")

# Create table
df_stats = pd.DataFrame([female_stats, male_stats])

# Round values
df_stats = df_stats.round(3)

# Display as a table
display(df_stats)
# ## 6. Add BMI (body mass index) as an 8th column to the female matrix
#
# Purpose: compute BMI = weight(kg) / (height(m))^2 for each female participant
# and append it as a new last column to the female numpy matrix.
#
# After the cell we print shape and a small preview.

# %%
# Height is column 1 in cm; convert to meters:
female_height_m = female[:, 1] / 100.0
female_weight_kg = female[:, 0]

# Compute BMI, handle any zero or NaN heights defensively:
with np.errstate(divide="ignore", invalid="ignore"):
    female_bmi = female_weight_kg / (female_height_m ** 2)

# Append as 8th column
female = np.column_stack([female, female_bmi])

print("female new shape (with BMI):", female.shape)
print("first 5 rows (weights, heights, ..., BMI):\n", female[:5, -1])  # only BMI column printed

# Summary:
# BMI values were computed and appended. They will be used in subsequent steps (standardization, ranking).


# %% [markdown]
# ## 7. Create standardised female dataset `zfemale` (z-scores of each column)
#
# Purpose: create a version of the female dataset where each column has been standardised to zero mean and unit standard deviation (z-scores).
# This is often useful for multivariate comparisons and visualisation.
#
# We will build a pandas DataFrame with column names for clarity and then standardise.

# %%
# Create DataFrame with clear column names before standardisation
# %% [markdown]
# ## 7. Create standardised female dataset zfemale

# %%

import numpy as np
import pandas as pd

# Column names
colnames = [
    "BMXWT",
    "BMXHT",
    "BMXARML",
    "BMXLEG",
    "BMXARMC",
    "BMXHIP",
    "BMXWAIST",
    "BMI"
]

# Create DataFrame for female data
female_df = pd.DataFrame(
    female,
    columns=colnames
)

# Calculate standardised z-scores
zfemale = (
    female_df - female_df.mean()
) / female_df.std(ddof=0)

# Create DataFrame for standardised values
zfemale_df = zfemale.copy()

# Display the shape
print("zfemale shape:", zfemale.shape)

# Display first 5 rows as a table
print("\nStandardised (z) values - first 5 females:")

display(
    zfemale_df.head().round(3)
)
# Summary:
# `zfemale` contains z-scores of all female variables, including the newly added BMI column.
# Means should be approximately zero and standard deviations approximately one for each column
# (subject to numerical rounding and any missing values).


# %% [markdown]
# ## 8. Scatterplot matrix (pairplot) for standardised female variables:
#
# We will visualise the pairwise relations among the **standardised** versions of:
# - height (`BMXHT`), weight (`BMXWT`), waist circumference (`BMXWAIST`), hip circumference (`BMXHIP`), and BMI (`BMI`).
#
# Purpose: explore pairwise relationships visually using seaborn.pairplot, and compute Pearson and Spearman correlation matrices for these variables.
#
# After the plots and correlation matrices we will provide an interpretation.

# %%
import seaborn as sns

# Choose columns of interest
vars_of_interest = ["BMXWT", "BMXHT", "BMXWAIST", "BMXHIP", "BMI"]
z_df_plot = zfemale_df[vars_of_interest].copy()

# Pairplot (scatterplot matrix)
sns.set(style="ticks", context="notebook")
pp = sns.pairplot(z_df_plot, diag_kind="kde", plot_kws={"s": 10, "alpha": 0.6})
pp.fig.suptitle("Pairplot of standardised female variables (WT, HT, Waist, Hip, BMI)", y=1.02)
plt.show()

# Compute Pearson and Spearman correlations
pearson_corr = z_df_plot.corr(method="pearson")
spearman_corr = z_df_plot.corr(method="spearman")
print("Pearson correlation matrix:")
display(pearson_corr.round(3))

print("Spearman correlation matrix:")
display(spearman_corr.round(3))


# Interpretation (brief):
# - Strong positive Pearson correlations are expected between weight and BMI, weight and hip/waist circumferences.
# - Height often correlates with weight moderately; BMI is strongly related to weight and less strongly (or negatively) with height.
# - Spearman correlations (rank-based) will be similar if relationships are monotonic; differences point to non-linearities or influential outliers.
#
# The printed matrices quantify these relationships. For example, a high Pearson correlation between BMI and weight indicates that heavier females have higher BMI (by construction), while correlations between waist and hip with weight indicate how body circumferences scale with body mass.


# %% [markdown]
# ## 9. Compute waist-to-height and waist-to-hip ratios for both sexes (add as columns)
#
# Purpose: compute the two ratios for male and female participants and append them as new columns
# to the `male` and `female` matrices. We will keep the existing ordering of columns; the new columns will be appended.
#
# Columns added:
# - waist_to_height = waist(cm) / height(cm)
# - waist_to_hip = waist(cm) / hip(cm)

# %%
def append_ratios(arr, is_female=False):
    # arr: numpy array with at least columns: HEIGHT at index 1, HIP index 5, WAIST index 6
    # If female==True then BMI already present at index 7 (but we won't rely on index for ratios)
    # Defensive conversion:
    arr = np.array(arr, dtype=float)
    height_cm = arr[:, 1]
    hip_cm = arr[:, 5]
    waist_cm = arr[:, 6]
    # compute ratios; handle division by zero
    with np.errstate(divide="ignore", invalid="ignore"):
        waist_to_height = waist_cm / height_cm
        waist_to_hip = waist_cm / hip_cm
    new = np.column_stack([arr, waist_to_height, waist_to_hip])
    return new

male = append_ratios(male, is_female=False)
female = append_ratios(female, is_female=True)

print("male new shape:", male.shape)
print("female new shape:", female.shape)

print("Female last 3 columns: BMI, waist/height, waist/hip")

female_ratio_table = pd.DataFrame(
    female[:5, -3:],
    columns=["BMI", "Waist/Height", "Waist/Hip"]
)

display(female_ratio_table.round(3))

# Summary:
# Both matrices now include two extra columns (waist-to-height and waist-to-hip ratios).
# These will be used for comparisons and boxplots below.


# %% [markdown]
# ## 10. Boxplot with four boxes comparing waist-to-height and waist-to-hip ratios for both sexes
#
# Purpose: create a boxplot with four boxes side by side:
# 1. Female waist-to-height
# 2. Female waist-to-hip
# 3. Male waist-to-height
# 4. Male waist-to-hip
#
# This enables direct comparison of the distributional properties of these two ratio measures across sexes.

# %%
# indices for appended columns:
# For female matrix, the appended ratios are the last two columns:
female_w2h = female[:, -2]
female_w2hip = female[:, -1]
male_w2h = male[:, -2]
male_w2hip = male[:, -1]

data_for_boxplot = [female_w2h, female_w2hip, male_w2h, male_w2hip]
labels = ["F: waist/height", "F: waist/hip", "M: waist/height", "M: waist/hip"]

plt.figure(figsize=(10, 6))
plt.boxplot(data_for_boxplot, tick_labels=labels, showfliers=True)
plt.ylabel("Ratio")
plt.title("Comparison of waist-to-height and waist-to-hip ratios (female vs male)")
plt.grid(axis="y", linestyle=":", alpha=0.5)
plt.xticks(rotation=20)
plt.show()

# Explanation:
# - The boxplot shows central tendency and spread for each ratio by sex.
# - Waist-to-height ratio typically falls around ~0.45–0.6 for adults (varies), and is often higher in males.
# - Waist-to-hip shows the relative waist compared to hip; values >1 indicate waist larger than hip (rare except for measurement or rare body shapes).
# - Inspect the medians and IQRs: whichever group shows larger median suggests systematically higher ratio for that group.
# - Outliers highlight individuals with extreme central adiposity or measurement extremes.


# %% [markdown]
# ## 11. Advantages and disadvantages of BMI, waist-to-height ratio, and waist-to-hip ratio
#
# Purpose: provide a short written summary (qualitative) of strengths and weaknesses of these measures.

# %%
from textwrap import dedent

discussion = dedent(
    """
    Advantages and disadvantages (concise):

    - BMI (Body Mass Index)
        Advantages:
          * Simple to compute from weight and height.
          * Widely used; provides a quick screening tool for underweight/overweight/obesity.
          * Good for population-level surveillance.
        Disadvantages:
          * Does not distinguish fat vs. lean mass (e.g., muscular individuals flagged as overweight).
          * Does not capture fat distribution (central vs peripheral).
          * Age, sex, and ethnicity differences are not accounted for in a raw BMI value.

    - Waist-to-height ratio (WHtR)
        Advantages:
          * Reflects central adiposity relative to body size; better predictor of cardiometabolic risk than BMI in many studies.
          * Simple to compute and interpretable (rule of thumb: keep waist < 0.5 × height).
        Disadvantages:
          * Still a simple ratio that ignores body composition details.
          * Cutoffs may vary by population; measurement technique matters.

    - Waist-to-hip ratio (WHR)
        Advantages:
          * Measures fat distribution (abdomen vs hips); high WHR indicates more central (visceral) fat.
          * Historically used in epidemiological studies linking fat distribution to disease risk.
        Disadvantages:
          * Requires two measurements (waist and hip) with potential measurement variability.
          * May be less intuitive and less consistent across body shapes than WHtR.
          * Less sensitive to absolute adiposity than BMI or WHtR (someone with low hip circumference could have high WHR even if not obese).
    """
)
print(discussion)

# Summary:
# The printed text contains a brief, balanced list of pros and cons for the three measures requested.


# %% [markdown]
# ## 12. Standardised body measurements for females with lowest and highest BMI
#
# Purpose: identify the 5 female participants with the lowest BMI and the 5 with the highest BMI,
# then print their standardized body measurements (z-scores) from `zfemale`.
#
# We will show the full row of standardised measurements (the same columns we used earlier).
# After printing, a short interpretation will follow.

# %%
# female_df currently has BMI in column 'BMI' (last)
female_bmi_vals = female_df["BMI"].values
order = np.argsort(female_bmi_vals)

lowest5_idx = order[:5]
highest5_idx = order[-5:]

selected_idx = np.concatenate([lowest5_idx, highest5_idx])
selected_idx_sorted = np.sort(selected_idx)  # for nicer printing, keep original relative order if desired

print("Indices of lowest 5 BMI (female):", lowest5_idx)
print("Indices of highest 5 BMI (female):", highest5_idx)
print("\nStandardised body measurements for these 10 females (columns: {})\n".format(", ".join(colnames)))
print(zfemale_df.iloc[np.concatenate([lowest5_idx, highest5_idx])].to_string(float_format="{:.3f}".format))

display(zfemale_df.iloc[np.concatenate([lowest5_idx, highest5_idx])].round(3))

# Interpretation:
# - The rows corresponding to lowest BMI participants will show strongly negative z-scores for BMI and often for weight,
#   but not necessarily for height (tall and light individuals may have low BMI with near-zero height z-score).
# - The highest BMI participants will typically show strongly positive z-scores for BMI and weight, and often for hip/waist.
# - The standardised table helps identify which other measurements (e.g., hip, waist) are unusually large/small relative to the female population.
#
# For example, a high-BMI person with large positive z-scores in waist and hip indicates central adiposity; a high-BMI person with moderate waist and low height z-score suggests another body composition pattern.

# %% [markdown]
# ## Final remarks
#
# - The notebook followed the assignment tasks step by step.
# - All necessary plots and tables are produced.
# - If you intend to rerun this script in VS Code, open it and run cells as you would in a Jupyter notebook.
# - If you need the output exported as a true `.ipynb` file, I can provide one; for VS Code interactive use the provided `.py` with cell markers.
#
# End of analysis.
```
